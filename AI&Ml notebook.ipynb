```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

# dummy data (replace with your real X_train, y_train)
X_train = np.random.rand(100, 10)
y_train = np.random.randint(0, 2, 100)

# model
model = Sequential([
    Dense(16, activation='relu', input_shape=(10,)),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# TRAIN (epochs = 10)
model.fit(X_train, y_train, epochs=10, verbose=1)

# SAVE MODEL
model.save("loan_model.keras")

print("MODEL SAVED SUCCESSFULLY")
```

```python
import gradio as gr
import numpy as np
from tensorflow.keras.models import load_model

model = load_model("loan_model.keras")

def predict(age, income, debt, loan):
    data = np.array([age, income, debt, loan] + [0]*6)  # adjust 10 inputs
    data = data.reshape(1, -1)

    pred = model.predict(data)[0][0]

    if pred > 0.5:
        return f"🔴 HIGH RISK "
    else:
        return f"🟢 LOW RISK "

app = gr.Interface(
    fn=predict,
    inputs=[
        gr.Number(label="Age"),
        gr.Number(label="Income"),
        gr.Number(label="Debt Ratio"),
        gr.Number(label="Loan Amount")
    ],
    outputs="text",
    title="Loan Risk Predictor"
)

app.launch()
```
